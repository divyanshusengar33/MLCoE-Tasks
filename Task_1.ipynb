```python
df["company_name"] = df["company_name"].str.strip()
df["job_role"] = df["job_role"].str.strip()
df["sector"] = df["sector"].str.strip()
df["company_type"] = df["company_type"].str.strip()
df["work_mode"] = df["work_mode"].str.strip()
```

```python
df["salary_min_usd"] = (
    df["salary_min_usd"]
    .astype(str)
    .str.replace("$", "")
    .str.replace(",", ""))
```

```python
df["salary_min_usd"] = pd.to_numeric(
    df["salary_min_usd"], errors="coerce")
```

converting  min salary into nymbers


```python
df["salary_max_usd"] = (
    df["salary_max_usd"].astype(str)
    .str.replace("$", "")
    .str.replace(",", ""))

```

```python
df["salary_max_usd"] = pd.to_numeric(
    df["salary_max_usd"], errors="coerce")
```

converting max salary into numbers

```python
print(df["salary_min_usd"].dtype)
print(df["salary_max_usd"].dtype)
```

```python
df
```

```python
print(df[df["salary_min_usd"] <= 0])
```

```python
df.loc[df["salary_min_usd"] <= 0, "salary_min_usd"] = np.nan
df.loc[df["salary_max_usd"] <= 0, "salary_max_usd"] = np.nan
```

```python
print(df["years_experience_min"].min())
print(df["years_experience_max"].max())
```

```python
df.loc[df["years_experience_min"] > df["years_experience_max"],["years_experience_min", "years_experience_max"]] = np.nan
```

minimum experience should not be greater than maximum experience

```python
df.loc[df["number_of_applicants"] < 0,"number_of_applicants"] = np.nan
```

```python
for col in ["posting_date", "application_deadline", "date_position_filled"]:
    df[col] = pd.to_datetime(df[col], errors="coerce")
```

converting date columns into proper date format

```python
cols = ["salary_min_usd", "salary_max_usd", "number_of_applicants", "number_of_interview_rounds", "offers_extended", "offers_accepted"]

df[cols] = df[cols].fillna(df[cols].median())
```

filling all the numerical value with median

```python
df["layoff_within_12_months_flag"] = df["layoff_within_12_months_flag"].map({'N':"No", '0' : "No", '1' : "Yes" ,'False':"No" ,'Y':"Yes", 'True':"Yes"})
df["visa_sponsorship_offered"] = df["visa_sponsorship_offered"].map({'N':"No", '0' : "No", '1' : "Yes" ,'False':"No" ,'Y':"Yes", 'True':"Yes"})

print(df["layoff_within_12_months_flag"].unique())
print(df["visa_sponsorship_offered"].unique())
```

```python
df["primary_programming_language"] = df["primary_programming_language"].fillna("unknown")
```

```python
df["job_location_city"] = df["job_location_city"].fillna("Unknown")
```

```python
print(df.isnull().sum())
```

###Feature engineering

```python
df["average_salary"] = (df["salary_min_usd"] +df["salary_max_usd"]) / 2
```

avg salary

```python
df["average_experience"] = (df["years_experience_min"] + df["years_experience_max"]) / 2
```

avg experience

```python
df["experience_range"] = (df["years_experience_max"] - df["years_experience_min"])
```

how much person have experience

```python
df["posting_year"] = df["posting_date"].dt.year
```

we can extract year in which job opening has been released

```python
df["posting_month"] = df["posting_date"].dt.month
```

###Bivariate Analysis

```python
print(
    df["average_experience"].corr(
        df["average_salary"]))
```

salary has a strong  positive corelation with experience, means  more experience offer higher salary


```python
sns.boxplot(
 data=df,
 x="seniority_level",
 y="average_salary")
plt.title("Salary by Seniority Level")
plt.xlabel("Seniority Level")
plt.ylabel("Average Salary")
plt.show()
```

salary generally increase as seniority level increases

```python
plt.figure(figsize=(8,6))
sns.countplot(data=df, x='sector', hue='layoff_within_12_months_flag')
plt.title('Layoffs by Sector')
plt.xticks(rotation=90)
plt.show()
```

```python
df.groupby("work_mode")["average_salary"].mean().plot(kind="bar")
plt.title("Average Salary by Work Mode")
plt.xlabel("Work Mode")
plt.ylabel("Average Salary")

plt.show()
```

```python
pd.crosstab(
df["seniority_level"],
df["work_mode"]).plot(kind="bar")
plt.title("Seniority Level vs Work Mode")
plt.xlabel("Seniority Level")
plt.ylabel("Number of Jobs")
plt.xticks(rotation=45)
plt.show()
```

this graph shows which seniority level has how many no. of jobs

###Multivariate Analysis

```python
table = pd.pivot_table(df,
values="average_salary",
index="seniority_level",
columns="work_mode",
aggfunc="mean")
print(table)
```

```python
plt.figure(figsize=(8,5))
sns.heatmap(
table,
annot=True,
fmt=".0f")
plt.title("Average Salary by Seniority and Work Mode")
plt.xlabel("Work Mode")
plt.ylabel("Seniority Level")
plt.show()
```

seniority has a much stronger effect on salry than work mode

Time Series Analysis

```python
df["posting_year"] = df["posting_date"].dt.year
```

```python
yearly_jobs = df.groupby("posting_year").size()
print(yearly_jobs)
```

```python
plt.figure(figsize=(9,5))
plt.plot(
yearly_jobs.index,
yearly_jobs.values,
marker="o")
plt.title("Job Postings Over Time")
plt.xlabel("Year")
plt.ylabel("Number of Job Postings")
plt.grid()
plt.show()
```

this shows that 2025 has higher no. of job posting than other

```python
monthly_jobs = df.groupby(df["posting_date"].dt.to_period("M")).size()
```

```python
plt.figure(figsize=(16,4))
plt.plot(
    monthly_jobs.index.astype(str),
    monthly_jobs.values)
plt.title("Monthly Job Posting Trend")
plt.xlabel("Month")
plt.ylabel("Number of Job Postings")
plt.xticks(rotation=90)
plt.grid()
plt.show()
```

monthly jobs posting show that how hiring changes throughout the year
